## 1)A) Sintetizar el cuadripolo pasivo sin pérdidas. Considerar Rg=1

$$ \boxed{T(S) = \frac{V_2}{V_g} = \frac{k·S(S^2 + \frac{1}{16})}{S^3 + 2·S^2 + 2·S + 1}} (1)$$

Ecuacion de parametros impedancia:

- $ V_1 = Z_{11}·I_1 + Z_{12}·I_2$
- $ V_2 = Z_{21}·I_1 + Z_{22}·I_2$

Malla de entrada:

- $ V_g = I_1·R_g + V_1 $

$$ V_1|_{I_2 = 0} = Z_{11}·I_1 \quad ; \quad V_2 = Z_{21}·I_1 \quad \implies \quad \boxed {I_1 = \frac{V_2}{Z_{21}}} \quad \implies \quad \boxed{V_1 = Z_{11}·\frac{V_2}{Z_{21}}} $$

$$ V_g = \frac{V_2}{Z_{21}}·R_g + Z_{11}·\frac{V_2}{Z_{21}} = V_2(\frac{R_g + Z_{11}}{Z_{21}}) \quad \implies \quad T(s) = \frac{V_2}{V_g} = \frac{Z_{21}}{R_g + Z_{11}} \quad y \quad R_g = 1 \quad \implies \quad \boxed{T(s) = \frac{Z_{21}}{Z_{11} + 1}}(2) $$

$$ (2) = (1) \quad \implies \quad \frac{Z_{21}}{Z_{11} + 1} = \frac{k·S(S^2 + \frac{1}{16})}{S^3 + 2·S^2 + 2·S + 1} = \frac{k·S(S^2 + \frac{1}{16})}{S^3 + 2·S + 2·S^2 + 1} = \frac{\frac{k·S(S^2 + \frac{1}{16})}{2·S^2 + 1}}{\frac{S^3 + 2·S}{2·S^2 + 1} + 1} $$

$$ \boxed{Z_{21} = \frac{k·S(S^2 + \frac{1}{16})}{2·S^2 + 1}} \quad y \quad \boxed{Z_{11} = \frac{S^3 + 2·S}{2·S^2 + 1}} $$

- La funcion inmitancia $Z_{11}$ es sobre la cual se realizan las remociones y la funcion $Z_{21}$ es la que nos impone las singularidades a remover

$$ Y_{11} = \frac{1}{Z_{11}} = \frac{2·S^2 + 1}{S^3 + 2·S} = \frac{2(S^2 + \frac{1}{2})}{S(S^2 + 2)} $$

Desplazo el Cero en $S^2 = -1 \,\,$ a $\,\, S^2 = -\frac{1}{16}$:

$$ Y_1 = Y_{11} - \frac{K^`}{S} \quad y \quad K^` = \lim_{S^2 \to -\frac{1}{16}} S·Y_{11}(S) $$

$$ K^` = \lim_{S^2 \to -\frac{1}{16}} S·\frac{2(S^2 + \frac{1}{2})}{S(S^2 + 2)} = \lim_{S^2 \to -\frac{1}{16}} \frac{2(S^2 + \frac{1}{2})}{S^2 + 2} = \frac{14}{31} \quad \implies \quad \boxed{K^` = \frac{14}{31}} \quad y \quad \boxed{L^` = \frac{1}{K^`} = 2.21} $$

- $L^`$ representa un inductor en derivacion

$$ Y_1 = Y_{11} - \frac{14}{31}·\frac{1}{S} = \frac{2(S^2 + \frac{1}{2})}{S(S^2 + 2)} - \frac{14}{31}·\frac{1}{S} = \frac{2·S^2 + 1 - \frac{14}{31}·(S^2 + 2)}{S(S^2 + 2)} = \frac{2·S^2 + 1 - \frac{14}{31}·S^2 - \frac{28}{31}}{S(S^2 + 2)} $$

$$ Y_1 = \frac{\frac{48}{31}·S^2 + \frac{3}{31}}{S(S^2 + 2)} = \frac{48}{31} \frac{S^2 + \frac{1}{16}}{S(S^2 + 2)} \quad \implies \quad Z_1 = \frac{1}{Y_1} = \frac{31}{48} \frac{S(S^2 + 2)}{S^2 + \frac{1}{16}} $$

Se remueve el polo de $Z_1$ en $S^2 = -\frac{1}{16} $ (este polo representa un tanque LC en serie):

$$ Z_2 = Z_1 - \frac{2·K_1·S}{S^2 + \frac{1}{16}} \quad y \quad 2·K_1 = \lim_{S^2 \to -\frac{1}{16}} \frac{S^2 + \frac{1}{16}}{S}·Z_1(S) $$

$$ 2·K_1 = \lim_{S^2 \to -\frac{1}{16}} \frac{S^2 + \frac{1}{16}}{S}·\frac{31}{48} \frac{S(S^2 + 2)}{S^2 + \frac{1}{16}} = \lim_{S^2 \to -\frac{1}{16}} \frac{31}{48}·(S^2 + 2) = \frac{31}{48}·(-\frac{1}{16} + 2) = \frac{31}{48}·\frac{31}{16} = \frac{961}{768} $$

$$ 2·K_1 = \frac{961}{768} \quad y \quad L_1 = \frac{2·K_1}{\omega_0^2} = \frac{961}{768}·\frac{1}{\frac{1}{16}} = \frac{961}{48} = 20.02 H \quad ; \quad C_1 = \frac{1}{2·K_1} = \frac{768}{961} = 0.8f $$

$$ \boxed{2·K_1 = \frac{961}{768}} \quad ; \quad  \boxed{L_1 = 20.02H} \quad ; \quad \boxed{C_1 = 0.8f} $$

$$ Z_2 = Z_1 - \frac{\frac{961}{768}·S}{S^2 + \frac{1}{16}} = \frac{31}{48} \frac{S(S^2 + 2)}{S^2 + \frac{1}{16}} - \frac{\frac{961}{768}·S}{S^2 + \frac{1}{16}} = \frac{31}{48}·S(\frac{S^2 + 2 - \frac{31}{16}}{S^2 + \frac{1}{16}}) = \frac{31}{48}·S(\frac{S^2 + \frac{1}{16}}{S^2 + \frac{1}{16}}) $$

$$ \boxed{Z_2 = \frac{31}{48}·S} $$

- $Z_2$ representa un inductor en serie de valor $ L_2 = \frac{31}{48} = 0.65H$

## 1)B) Hallar el valor de k que cumple con la síntesis y valor de los componentes hallados.

- Teoricamente:

$$ \lim_{S^2 \to \infty} T(S) = \lim_{S^2 \to \infty} \frac{k·S^3 + \frac{1}{16}·S}{S^3 + 2·S^2 + 2·S + 1} = K $$

- Fisicamente los inductores son circuitos abiertos y los capacitores son corto circuitos:



El circuito resultante es un circuito abierto, entonces $V_g = V_2 \quad \implies \quad T(s) = \frac{V_2}{V_g} = 1$

## 2)A) Sintetizar el cuadripolo pasivo sin pérdidas. Se considera RL = 1

$$ \boxed{T(S) = \frac{V_2}{I_1} = \frac{k·S(S^2 + 9)}{S^3 + 2·S^2 + 2·S + 1}}$$

Ecuacion de parametros impedancia:

- $ V_1 = Z_{11}·I_1 + Z_{12}·I_2$
- $ V_2 = Z_{21}·I_1 + Z_{22}·I_2$

Malla de Salida:

- $ V_2 = -V_{R_L} \quad \implies \quad V_2 = -I_2·R_L \quad \implies \quad I_2 = -\frac{V_2}{R_L}$

$$ V_2 = Z_{21}·I_1 + Z_{22}·(-\frac{V_2}{R_L}) \quad \implies \quad Z_{21}·I_1 = V_2(1 + \frac{Z_{22}}{R_L}) \quad \implies \quad T(S) = \frac{V_2}{I_1} = \frac{Z_{21}}{\frac{Z_{22}}{R_L} + 1} $$

$$ R_L = 1 \quad \implies \quad T(S) =  \frac{Z_{21}}{Z_{22} + 1} = \frac{k·S(S^2 + 9)}{S^3 + 2·S^2 + 2·S + 1} = \frac{k·S(S^2 + 9)}{S^3 + 2·S + 2·S^2 + 1} = \frac{\frac{k·S(S^2 + 9)}{2·S^2 + 1}}{\frac{S^3 + 2·S}{2·S^2 + 1} + 1} $$

$$ \boxed{Z_{21} = \frac{k·S(S^2 + 9)}{2·S^2 + 1}} \quad y \quad \boxed{Z_{22} = \frac{S^3 + 2·S}{2·S^2 + 1}} $$

- La funcion inmitancia $Z_{22}$ es sobre la cual se realizan las remociones y la funcion $Z_{21}$ es la que nos impone las singularidades a remover

$$ Z_1 = Z_{22} - K^`·S \quad y \quad K^` = \lim_{S^2 \to -9} \frac{1}{S}·Z_{22} = \lim_{S^2 \to -9} \frac{1}{S}·\frac{S^3 + 2·S}{2·S^2 + 1} = \lim_{S^2 \to -9} \frac{S^2 + 2}{2·S^2 + 1} = \frac{7}{17} $$

$$ \boxed{L^` = K^` =  \frac{7}{17}} $$

$$ Z_1 = \frac{S^3 + 2·S}{2·S^2 + 1} - \frac{7}{17}·S = \frac{S^3 + 2·S - \frac{7}{17}·S(2·S^2 + 1)}{2·S^2 + 1} = \frac{S^3 + 2·S - \frac{14}{17}·S^3 + \frac{7}{17}·S}{2·S^2 + 1} $$

$$ Z_1 = \frac{\frac{3}{17}·S^3 + \frac{27}{17}·S}{2·S^2 + 1} \quad \implies \quad \boxed{Z_1 = \frac{3}{17}\frac{S(S^2 + 9)}{2·S^2 + 1}} \quad y \quad \boxed{Y_1 = \frac{17}{3}\frac{2·S^2 + 1}{S(S^2 + 9)}} $$

Remocion total del polo en $S^2 = -9$:

$$ Y_2 = Y_1 - \frac{2·K_1·S}{S^2 + 9} \quad y \quad 2·K_1 = \lim_{S^2 \to -9} \frac{S^2 + 9}{S}·Y_1 $$

$$ 2·K_1 = \lim_{S^2 \to -9} \frac{S^2 + 9}{S}·\frac{17}{3}\frac{2·S^2 + 1}{S(S^2 + 9)} = \lim_{S^2 \to -9} \frac{17}{3}\frac{2·S^2 + 1}{S^2} = \lim_{S^2 \to -9} \frac{17}{3}(2 + \frac{1}{S^2}) = \frac{17}{3}(2 - \frac{1}{9}) = \frac{289}{27} $$

$$ \boxed{2·K_1 = \frac{289}{27}} \quad y \quad \boxed{L_1 = \frac{2·K_1}{\omega_o^2} = \frac{289}{27}·\frac{1}{9} = \frac{289}{243} = 1.189H} \quad y \quad \boxed{C_1 = \frac{1}{2·K_1} = \frac{27}{289} = 93.43mf} $$

$$ Y_2 = Y_1 - \frac{\frac{289}{27}·S}{S^2 + 9} = \frac{17}{3}\frac{2·S^2 + 1}{S(S^2 + 9)} - \frac{\frac{289}{27}·S}{S^2 + 9} = \frac{17}{3}\frac{2·S^2 + 1}{S(S^2 + 9)} - \frac{\frac{289}{27}·S^2}{S(S^2 + 9)}$$

$$ Y_2 = \frac{\frac{34}{3}·S^2 + \frac{17}{3} - \frac{289}{27}·S^2}{S(S^2 + 9)} = \frac{\frac{17}{27}·S^2 + \frac{17}{3}}{S(S^2 + 9)} = \frac{17}{27}\frac{S^2 + 9}{S(S^2 + 9)} = \frac{17}{27}\frac{1}{S} $$

- Queda un inductor en derivacion $\boxed{L_2 = \frac{27}{17}}$

### Sintesis del circuito:



## 2)B) Hallar el valor de k que cumple con la síntesis y valor de los componentes hallados.

$$ \boxed{T(S) = \frac{V_2}{I_1} = \frac{k·S(S^2 + 9)}{S^3 + 2·S^2 + 2·S + 1}}$$

- Para calcular el valor K hay que calcular $ \frac{V_2}{I_1}|_{I_2 = 0}$:



$$ I_1 = I_{L1} + I_{L2} \quad y \quad V_1 = V_2 \quad ; \quad I_{L1} = \frac{V_2}{SL_1 + \frac{1}{SC_1}} \quad y \quad I_{L2} = \frac{V_2}{SL_2} $$

$$ I_1 = \frac{V_2}{SL_1 + \frac{1}{SC_1}} + \frac{V_2}{SL_2} = (\frac{1}{\frac{S^2L_1C_1 + 1}{SC_1}} + \frac{1}{SL_2})V_2 = (\frac{SC_1}{S^2L_1C_1 + 1} + \frac{1}{SL_2})V_2 $$

$$ I_1 = \frac{S^2C_1L_2 + S^2L_1C_1 + 1}{SL_2(S^2L_1C_1 + 1)}V_2 = \frac{S^2C_1(L_1 + L_2) + 1}{S^3L_1L_2C_1 + L_2S}V_2 \quad \implies \quad Z_{21} = \frac{V_2}{I_1} = \frac{S^3L_1L_2C_1 + L_2S}{S^2C_1(L_1 + L_2) + 1} $$

$$ Z_{21} = \frac{L_1·L_2·C_1·S(S^2 + \frac{1}{L_1·C_1})}{S^2C_1(L_1 + L_2) + 1} \quad ; \quad \frac{1}{L_1·C_1} = 9 \quad \implies \quad L_1·C_1 = \frac{1}{9} \quad y \quad L_2 = \frac{27}{17} \quad \implies \quad L_1·L_2·C_1 = \frac{3}{17} $$

$$ C_1(L_1 + L_2) = 2 \quad \implies \quad Z_{21} = \frac{\frac{3}{17}·S(S^2 + 9)}{2·S^2 + 1} \quad \implies \quad \boxed{K = \frac{3}{17}} $$

## 2) Para la Red T:

### A) Obtener $\frac{V_{23}}{V_{13}} \quad ; \quad \frac{V_{21}}{V_{31}} \quad y \quad \frac{V_{12}}{V_{32}} \quad$, explicitando el armado de los cofactores de segundo orden.

- Calculo de $\frac{V_{23}}{V_{13}}$:

$$
\frac{V_{23}}{V_{13}} = \frac{\Delta_{12}}{\Delta_{11}} \quad \implies \quad
Y_{(3,3)} = 
\begin{bmatrix}
\frac{1}{SL} & 0 & -\frac{1}{SL} \\
0 & \frac{1}{SL} & -\frac{1}{SL} \\
-\frac{1}{SL} & -\frac{1}{SL} & SC + \frac{2}{SL}
\end{bmatrix}
$$

$$
\Delta_{11} = det
\begin{bmatrix}
\frac{1}{SL} & -\frac{1}{SL} \\
-\frac{1}{SL} & SC + \frac{2}{SL}
\end{bmatrix} = \frac{1}{SL} (SC + \frac{2}{SL}) - (-\frac{1}{SL})(-\frac{1}{SL}) = \frac{C}{L} + 2·(\frac{1}{SL})^2 - (\frac{1}{SL})^2 = \frac{C}{L} + (\frac{1}{SL})^2
$$

$$ \Delta_{11} = \frac{S^2LC + 1}{(SL)^2} $$

$$
\Delta_{12} = -det
\begin{bmatrix}
0 & -\frac{1}{SL} \\
-\frac{1}{SL} & SC + \frac{2}{SL}
\end{bmatrix} = -(0 ·(SC + \frac{2}{SL}) - (-\frac{1}{SL})(-\frac{1}{SL})) =  (\frac{1}{SL})^2 
$$

$$ \Delta_{12} = (\frac{1}{SL})^2 $$

$$ \boxed{\frac{V_{23}}{V_{13}} = \frac{(\frac{1}{SL})^2}{\frac{SC + 1}{(SL)^2}} = \frac{1}{SC + 1}} $$

- Calculo de $\frac{V_{21}}{V_{31}}$:

$$
\frac{V_{21}}{V_{31}} = \frac{\Delta_{21}}{\Delta_{22}} \quad \implies \quad
Y_{(3,3)} = 
\begin{bmatrix}
\frac{1}{SL} & 0 & -\frac{1}{SL} \\
0 & SC & -SC \\
-\frac{1}{SL} & -SC & SC + \frac{2}{SL}
\end{bmatrix}
$$

$$
\Delta_{21} = -det
\begin{bmatrix}
0 & -\frac{1}{SL} \\
-SC & SC + \frac{2}{SL}
\end{bmatrix} = -(0 · (SC + \frac{2}{SL}) - (-\frac{1}{SL})(-SC))
$$

$$ \Delta_{21} = \frac{C}{L} $$

$$
\Delta_{22} = det
\begin{bmatrix}
\frac{1}{SL} & -\frac{1}{SL} \\
-\frac{1}{SL} & SC + \frac{2}{SL}
\end{bmatrix} = \frac{1}{SL} · (SC + \frac{2}{SL}) - (-\frac{1}{SL})(-\frac{1}{SL}) = \frac{C}{L} + 2 (\frac{1}{SL})^2 - (\frac{1}{SL})^2
$$

$$ \Delta_{22} = \frac{C}{L} + (\frac{1}{SL})^2 = \frac{S^2LC + 1}{(SL)^2} $$

$$ \boxed{\frac{V_{21}}{V_{31}} = \frac{\frac{C}{L}}{\frac{S^2LC + 1}{(SL)^2}} = \frac{S^2LC}{S^2LC + 1}} $$

- Calculo de $\frac{V_{12}}{V_{32}}$:

$$
\frac{V_{12}}{V_{32}} = \frac{\Delta_{21}}{\Delta_{22}} \quad \implies \quad
Y_{(3,3)} = 
\begin{bmatrix}
\frac{1}{SL} & 0 & -\frac{1}{SL} \\
0 & SC & -SC \\
-\frac{1}{SL} & -SC & SC + \frac{2}{SL}
\end{bmatrix}
$$

$$
\Delta_{21} = -det
\begin{bmatrix}
0 & -\frac{1}{SL} \\
-\frac{1}{SL} & SC + \frac{2}{SL}
\end{bmatrix} = -(0 · (SC + \frac{2}{SL}) - (-\frac{1}{SL})(-SC))
$$

$$ \Delta_{21} =  \frac{C}{L}$$

$$
\Delta_{22} = det
\begin{bmatrix}
\frac{1}{SL} & -\frac{1}{SL} \\
-\frac{1}{SL} & SC + \frac{2}{SL}
\end{bmatrix} = \frac{1}{SL}·(SC + \frac{2}{SL}) - (-\frac{1}{SL})(-\frac{1}{SL}) = \frac{C}{L} + 2 (\frac{1}{SL})^2 - (\frac{1}{SL})^2
$$

$$ \Delta_{22} = \frac{C}{L} + (\frac{1}{SL})^2 = \frac{S^2LC + 1}{(SL)^2} $$

$$ \boxed{\frac{V_{12}}{V_{32}} = \frac{\frac{C}{L}}{\frac{S^2LC + 1}{(SL)^2}} = \frac{S^2LC}{S^2LC + 1}} $$

## Para el MOS:

### A) Dada la MAI hallada en 1 obtener los parámetros Y para las configuraciones source común y drain común.

- #### Matriz de parametros Y para la configuracion Source Común (Nodo 3 referencia)

$$
Y_{Source\_Común} = 
\begin{bmatrix}
SC_{ISS} & -SC_{RSS} \\
g_m - SC_{RSS} & SC_{RSS} + \frac{1}{r_0} 
\end{bmatrix}
$$

- #### Matriz de parametros Y para la configuracion Drain Común (Nodo 2 referencia)

$$
Y_{Source\_Común} = 
\begin{bmatrix}
SC_{ISS} & -S(C_{ISS} - C_{RSS}) \\
-g_m - S(C_{ISS} - C_{RSS}) & S(C_{ISS} - C_{RSS}) + \frac{1}{r_0} + g_m
\end{bmatrix}
$$

### B) Calcular la matriz de parámetros ABCD en configuración source común

$$ A = -\frac{Y_{22}}{Y_{21}} = -\frac{Sr_0C_{RSS} + 1}{r_0(g_m - SC_{RSS})} \quad \implies \quad \boxed{A = \frac{Sr_0C_{RSS} + 1}{Sr_0C_{RSS} - g_mr_0}} $$

$$ B = -\frac{1}{Y_{21}} \quad \implies \quad \boxed{B = \frac{1}{SC_{RSS} - g_m}} $$

$$ C = -\frac{\Delta Y}{Y_{21}} \quad y \quad \Delta Y = SC_{ISS} · (SC_{RSS} + \frac{1}{r_0}) - (-SC_{RSS})(g_m - C_{RSS}) = S^2C_{RSS}·(C_{ISS} - C_{RSS}) + S (\frac{C_{ISS}}{r_0} + g_mC_{RSS}) $$

$$ \boxed{ C = \frac{S^2C_{RSS}·(C_{ISS} - C_{RSS}) + S (\frac{C_{ISS}}{r_0} + g_mC_{RSS})}{SC_{RSS} - g_m}} $$

$$ D = -\frac{Y_{11}}{Y_{21}} = - \frac{SC_{ISS}}{g_m - SC_{RSS}} \quad \implies \quad \boxed{ D = \frac{SC_{ISS}}{SC_{RSS} - g_m}} $$

$$
T = 
\begin{bmatrix}
A & B \\
C & D
\end{bmatrix} =
\begin{bmatrix}
\frac{Sr_0C_{RSS} + 1}{Sr_0C_{RSS} - g_mr_0} & \frac{1}{SC_{RSS} - g_m} \\
\frac{S^2C_{RSS}·(C_{ISS} - C_{RSS}) + S (\frac{C_{ISS}}{r_0} + g_mC_{RSS})}{SC_{RSS} - g_m} & \frac{SC_{ISS}}{SC_{RSS} - g_m}
\end{bmatrix}
$$

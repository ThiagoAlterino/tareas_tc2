## A) Obtener la transferencia pasa alto

Para obtener el orden del filtro se hay que encontrar el equivalente pasa bajo: $\quad f_P = 1 \quad $ y $\quad f_S = 3 \quad$

$$\xi^2 = 10^{\frac{\alpha_{MAX}}{10}} - 1  = 0.259 \quad y \quad \alpha_{n} = 10dB·Log(1+\xi^2·cosh^2(n·cosh^{-1}(f_S)))$$

$$n = 1 \quad \implies \quad \alpha_{1} = 5.23 dB \quad ; \quad n = 2 \quad \implies \quad \alpha_{2} = 18.8 dB \quad ; \quad n = 3 \quad \implies \quad \boxed{\alpha_{3} = 34 dB}$$

El orden del filtro debe ser 3
$$T(S)·T(-S) = \frac{1}{1 + \xi^2(C_3(\omega))^2} \quad ; \quad C_3(\omega) = 2·\omega·C_2 - C_1 \quad y \quad C_0 = 1 \quad y \quad C_1 = \omega \quad \implies \quad C_2 = 2·\omega·C_1 - C_0 = 2·\omega·\omega - 1 = 2\omega^2 -1 $$
$$C_3 = 2·\omega·(2\omega^2 -1) - \omega = 4 \omega^3 - 2\omega -\omega \quad \implies \quad \boxed{C_3 = 4\omega^3 - 3\omega}$$

$$T(S)·T(-S) = \frac{1}{1 + \xi^2(C_3(\omega))^2} = \frac{1}{1 + \xi^2(4\omega^3 - 3\omega)^2} = \frac{\frac{1}{\xi^2}}{\frac{1}{\xi^2} + 16\omega^6 - 24\omega^4 + 9\omega^2 }= \frac{\frac{1}{16\xi^2}}{\omega^6 - \frac{24}{16}\omega^4 + \frac{9}{16}\omega^2 + \frac{1}{16\xi^2}}$$

$$\omega = \frac{S}{j} \quad \implies \quad T(S)·T(-S) = \frac{\frac{1}{16\xi^2}}{- S^6 - \frac{24}{16}S^4 - \frac{9}{16}S^2 + \frac{1}{16\xi^2}} = \frac{-\frac{1}{16\xi^2}}{S^6 + \frac{24}{16}S^4 + \frac{9}{16}S^2 - \frac{1}{16\xi^2}}$$

$$T(S)·T(-S) = \frac{-\frac{1}{16\xi^2}}{(S+0.494)(S^2 + 0.494 · S + 0.994)(S-0.494)(S^2 - 0.494 · S + 0.994)}$$

$$T(S) = \frac{\sqrt{\frac{1}{16\xi^2}}}{(S+0.494)(S^2 + 0.494 · S + 0.994)}$$

### Transformacion pasa alto:
$$T($) = T(S)|_{S = \frac{1}{$}} = \frac{0.491}{(\frac{1}{$}+0.494)(\frac{1}{$^2} + 0.494 · \frac{1}{$} + 0.994)} = \frac{0.491}{\frac{1}{$}(1 + 0.494 · $)\frac{1}{$^2}(1 + 0.494 · $ + 0.994·$^2)} = \frac{0.491 · $^3}{0.494($ + \frac{1}{0.494})·0.994·($^2 + \frac{0.494}{0.994} · $ + \frac{1}{0.994})}$$

$$T($) = \frac{$^3}{($ + 2.024)($^2 + 0.497· $ + 1.006)}$$

### Filtro pasa alto Chebyshev:
$$\boxed{T(S) = \frac{S^2}{S^2 + 0.497 · S + 1.006}·\frac{S}{S + 2.024}}$$

## B) Diagrama de polos y ceros, esquema cualitativo de la respuesta de modulo y fase y calculo del retardo de grupo

Polos y ceros:
$$T(S) = \frac{S^2}{S^2 + 0.497·S+1.006} · \frac{S}{S+2.024}$$
![Polos_Ceros](./archivos/Polos_Ceros.png)

Modulo y fase:
$$T(\omega) = T(S)|_{S=j·\omega} = \frac{(j\omega)^3}{(j\omega)^3 + 2.521(j\omega)^2 + 2.012(j\omega) + 2.036} = \frac{-j\omega^3}{(2.036 - 2.521·\omega^2) + j(2.012·\omega - \omega^3)}$$

$$T(\omega) = \frac{\omega^3}{\sqrt{(2.036 - 2.521·\omega^2)^2 + (2.012·\omega - \omega^3)^2}}e^{j(\frac{\pi}{2} - arctg(\frac{2.012·\omega - \omega^3}{2.036 - 2.521·\omega^2}))}$$

Modulo:
$$\boxed{|T(\omega)| = \frac{\omega^3}{\sqrt{(2.036 - 2.521·\omega^2)^2 + (2.012·\omega - \omega^3)^2}}}$$

Fase:
$$\boxed{\phi(\omega) = \frac{\pi}{2} - arctg(\frac{2.012·\omega - \omega^3}{2.036 - 2.521·\omega^2})}$$

![Respuesta_Modulo_Fase](./archivos/Respuesta_Modulo_Fase.png)

Expresion del retardo de grupo:
$$D(\omega) = -\frac{d\phi(\omega)}{d\omega} \quad \implies \quad D(\omega) = -\frac{d}{d\omega}(\frac{\pi}{2} - arctg(\frac{2.012·\omega - \omega^3}{2.036 - 2.521·\omega^2})) = \frac{d}{d\omega}(arctg(\frac{2.012·\omega - \omega^3}{2.036 - 2.521·\omega^2}))$$

$$\boxed{D(\omega) = \frac{2.521·\omega^4 - 1.036·\omega^2 + 4.096}{\omega^6 + 2.331·\omega^4 - 6.217·\omega^2 + 4.145}}$$

## D) Imoplementar la transferencia de C) con un circuito pasivo


Para implementar la etapa de orden 2 $\,\,T_2(S) = \frac{S^2 + 0.25}{S^2 + 0.497 S + 1.006}$, se implementara el siguiente circuito pasivo:

![pasivo_orden2](./archivos/pasivo_orden2.png)

La transferencia de este circuito es:
$$T(S) = \frac{S^2 + \frac{1}{L_2C_2}}{S^2 + \frac{R_1}{L_2} S + \frac{C_1 + C_2}{L_2C_1C_2}} \quad \implies \quad \frac{1}{L_2C_2} = 0.25 \quad y \quad \frac{R_1}{L_2} = 0.497 \quad y \quad \frac{C_1 + C_2}{L_2C_1C_2} = 1.006$$

Para implementar la etapa de orden 1 $\,\,T_1(S) = \frac{S}{S + 2.024}$, se implementara el siguiente circuito pasivo:

![pasivo_orden1](./archivos/pasivo_orden1.png)

La transferencia de este circuito es:
$$T(S) = \frac{S}{S + \frac{1}{R_1C_1}} \quad \implies \quad \frac{1}{R_1C_1} = 2.024$$

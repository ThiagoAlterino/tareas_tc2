A) Obtenga la función transferencia normalizada $T_N(s)$ del filtro pasa-bajo respecto al retardo normalizado, factorizada en secciones de orden no mayor a 2 (bicuad y bilineal), de modo que la ganancia en continua sea unitaria ($\vert{}T(0)\vert{} = 1$). Desnormalice la transferencia $T(s)$ en función de la frecuencia angular real $\omega$ utilizando el valor de $D_0$ especificado.

B) Obtenga el diagrama de polos y ceros de $T_N(s)$. Grafique cualitativamente:

    2) La fase.
    
    3) El retardo de grupo D(w).
    
Deduzca la expresión analítica formal del retardo de grupo $D(\omega) = -\frac{d\theta(\omega)}{d\omega}$ en función de $\omega$.

C) Para atenuar una interferencia de alta frecuencia sin alterar significativamente el comportamiento del retardo en la banda de paso, se propone agregar un cero de transmisión en el eje imaginario en $\omega_z = 100\text{ krad/s}$.

D) Implemente una red pasiva RLC cargada en ambos extremos (o terminada en resistencia) que sintetice la transferencia $T(s)$ obtenida en el punto A).

F) Escale la red implementada en el punto e) de forma tal que opere con el retardo real especificado ($D_0 = 50\ \mu\text{s}$) y adapte el nivel de impedancia a $R_0 = 1\text{ k}\Omega$.

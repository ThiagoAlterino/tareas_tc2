## Parte B: Trabajo de Investigación sobre NANO VNA

- Un analizador de redes vectorial (VNA, por sus siglas en inglés Vector Network Analyzer) es un instrumento de medición electrónico que se utiliza para caracterizar el comportamiento de componentes y circuitos de radiofrecuencia (RF), microondas y ondas milimétricas.

- Comparacion con un osciloscopio digital GW Instek:

| Característica | NanoVNA (H / H4) | DSO GW Instek (ej. GDS-1054B / 1102B) |
| :--- | :--- | :--- |
| **Ancho de Banda** | **50 kHz a 900 MHz** (extensible hasta 1,5 GHz mediante armónicos). No mide continua (DC). | **DC (0 Hz) a 50 MHz – 100 MHz** (según modelo). |
| **Rango Dinámico** | **$\approx 70\text{ dB}$** (hasta 300 MHz) y **$\approx 50\text{ dB}$** (hasta 900 MHz), gracias a su receptor heterodino de filtro angosto que rechaza el ruido fuera de banda. | **$\approx 45\text{ dB}$ a $48\text{ dB}$** teóricos, limitado por la resolución vertical de su conversor A/D de **8 bits** ($20\log_{10}(2^8) \approx 48{,}16\text{ dB}$) y el ruido de banda ancha. |
| **Impedancia de Entrada** | **$50\,\Omega$** (adaptada a líneas de transmisión y sistemas de RF para evitar reflexiones en los puertos SMA). | **$1\,\text{M}\Omega \parallel 16\text{ pF}$** (alta impedancia para no cargar circuitos de baja frecuencia; requiere puntas atenuadoras $\times 10$). |

- El NanoVNA cuenta con mediciones en el dominio del tiempo mediante la función TDR (Time Domain Reflectometry). A diferencia de un reflectómetro pulsado que inyecta un escalón de tensión real, mide la respuesta compleja en frecuencia ($S_{11}(f)$) a lo largo de todo su barrido y aplica matemáticamente una Transformada Inversa de Fourier (IFFT) para sintetizar la respuesta temporal al impulso o al escalón.

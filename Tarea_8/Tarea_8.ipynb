$$ K_0 = \lim_{S \to 0}S·F(S) = \lim_{S \to 0} S · \frac{S^4 + 4·S^2 + 3}{S(S^2 + 2)} = \lim_{S \to 0} \frac{S^4 + 4·S^2 + 3}{S^2 + 2} = \frac{3}{2} \quad \implies \quad K_0 = \frac{3}{2} \quad y \quad C_0 = \frac{1}{K_0} \quad \implies \quad \boxed{C_0 = \frac{2}{3}f} $$

$$ F_{R1}(S) = F(S) - \frac{1}{C_0·S} = \frac{S^4 + 4·S^2 + 3}{S(S^2 + 2)} - \frac{3}{2}·\frac{1}{S} = \frac{S^4 + 4·S^2 + 3-\frac{3}{2}(S^2 + 2)}{S(S^2 + 2)} = \frac{S^4 + 4·S^2 + 3 -\frac{3}{2}S^2 - 3 }{S(S^2 + 2)} $$

$$ F_{R1}(S) = \frac{S^4 + \frac{5}{2}·S^2 }{S(S^2 + 2)} = \frac{S^2(S^2 + \frac{5}{2}) }{S(S^2 + 2)} = \frac{S(S^2 + \frac{5}{2}) }{S^2 + 2} $$

$$ K_{\infty} = \lim_{S \to \infty}\frac{F_{R1}(S)}{S} = \lim_{S \to \infty} \frac{S(S^2 + \frac{5}{2}) }{S^2 + 2}·\frac{1}{S} = \lim_{S \to \infty} \frac{S^2 + \frac{5}{2}}{S^2 + 2} = 1 \quad y \quad L_{\infty} = K_{\infty} \quad \implies \quad \boxed{L_{\infty} = 1 H} $$

$$ F_{R2}(S) = F_{R1}(S) - S = \frac{S(S^2 + \frac{5}{2})}{S^2 + 2} - S = \frac{S(S^2 + \frac{5}{2}) - S(S^2 + 2)}{S^2 + 2} = \frac{S^3 + \frac{5}{2}·S - S^3 - 2·S}{S^2 + 2} $$

$$ F_{R2}(S) = \frac{\frac{1}{2}·S}{S^2 + 2}$$

- La expresion del tanque LC es: $\frac{2·K_1·S}{S^2 + \omega_1^2}$ y $F_{R2}(S)$ es el tanque, entonces:

$$ \frac{2·K_1·S}{S^2 + \omega_1^2} = \frac{\frac{1}{2}·S}{S^2 + 2} \quad \implies \quad 2·K_1 = \frac{1}{2} \quad y \quad \omega_1^2 = 2 $$

$$ C_1 = \frac{1}{2·K_1} = \frac{1}{\frac{1}{2}} = 2f \quad y \quad L_1 = \frac{2·K_1}{\omega_1^2} = \frac{\frac{1}{2}}{2} = \frac{1}{4}H $$

## 3) Encuentre el valor de los elementos que integran el siguiente circuito y que satisface la función impedancia propuesta:



$$ Z(S) = \frac{S^2 + 6·S + 8}{S^2 + 6·S + 5} $$

- La expresion teorica de los bloques series es:

$$ \boxed{Z(S) = R_1 + \frac{R_2·S}{S + \frac{R_2}{L_2}} + \frac{\frac{1}{C_1}}{S + \frac{1}{R_3·C_1}}} $$

- Expresion de la impedancia en fracciones simples:

$$ Z(S) = \frac{S^2 + 6·S + 8}{S^2 + 6·S + 5} = \frac{S^2 + 6·S + 5 + 3}{S^2 + 6·S + 5} = 1 + \frac{3}{S^2 + 6·S + 5} $$

$$ \frac{3}{S^2 + 6·S + 5} = \frac{3}{(S + 5)(S + 1)} = \frac{A}{S + 5} + \frac{B}{S + 1} = \frac{A(S + 1) + B(S + 5)}{(S + 5)(S + 1)} = \frac{(A + B)S + A +5B}{(S + 5)(S + 1)} $$

- $ A + B = 0 \quad \implies \quad A = -B$

- $ A + 5B = 3 \quad \implies \quad -B + 5B = 3 \quad \implies \quad 4B = 3 \quad \implies \quad B = \frac{3}{4} \quad \implies \quad A = - \frac{3}{4} $

- Expresion final: $ \quad \boxed{Z(S) = 1 + \frac{-\frac{3}{4}}{S + 5} + \frac{\frac{3}{4}}{S + 1}} $

- Para igualar las expresiones teoricas y la expresion final, hay que operar algebraica el termino del tanque RL para podes igualar termino a termino:

$$ \frac{R_2·S}{S + \frac{R_2}{L_2}} = \frac{R_2·(S + \frac{R_2}{L_2} - \frac{R_2}{L_2})}{S + \frac{R_2}{L_2}} = \frac{R_2·(S + \frac{R_2}{L_2})}{S + \frac{R_2}{L_2}} - \frac{R_2·\frac{R_2}{L_2}}{S + \frac{R_2}{L_2}} = R_2 - \frac{R_2·\frac{R_2}{L_2}}{S + \frac{R_2}{L_2}} $$

- La expresion teorica final de los bloques series es: $ \boxed{Z(S) = R_1 + R_2 - \frac{R_2·\frac{R_2}{L_2}}{S + \frac{R_2}{L_2}} + \frac{\frac{1}{C_1}}{S + \frac{1}{R_3·C_1}}} $

- Al igualar las expresiones teorica y la expresion final obtenida quedaria:

    - $ R_1 + R_2 = 1 \quad \implies \quad R_1 = 1 - R_2 = 1 - \frac{3}{20} = \frac{17}{20} \quad \implies \quad \boxed{R_1 =\frac{17}{20}} $
    - $ \frac{R_2}{L_2} = 5 \quad \implies \quad \frac{1}{L_2}·\frac{3}{20} = 5 \quad \implies \quad \boxed{L_2 = \frac{3}{100} = 0.03} $
    - $ R_2 \frac{R_2}{L_2} = \frac{3}{4} \quad \implies \quad R_2 · 5 = \frac{3}{4} \quad \implies \quad \boxed{R_2 = \frac{3}{20} = 0.15} $
    - $ \frac{1}{C_1} = \frac{3}{4} \quad \implies \quad \boxed{C_1 = \frac{4}{3}} $
    - $ \frac{1}{R_3C_1} = 1 \quad \implies \quad R_3 = \frac{1}{C_1} = \frac{3}{4} \quad \implies \quad \boxed{R_3 = \frac{3}{4}} $

- Valores finales:

$$ \boxed{R_1 = 0.85} \quad ; \quad \boxed{L_2 = 0.03} \quad ; \quad \boxed{R_2 = 0.15} \quad ; \quad \boxed{C_1 = 1.33} \quad ; \quad \boxed{R_3 = 0.75} $$

## 1) A) Hallar la transferencia H(z)



$$ Y(Z) = A(Z) + A·Y(Z)·Z^{-1} \quad y \quad A(Z) = X(Z) - Z^{-1}·X(Z) $$

$$ A(Z) = Y(Z) - A·Y(Z)·Z^{-1}) \quad \implies \quad A(Z) = Y(Z)(1 - A·Z^{-1}) \quad \implies \quad Y(Z)(1 - A·Z^{-1}) = X(Z)(1 - Z^{-1}) $$

$$ H(Z) = \frac{Y(Z)}{X(Z)} = \frac{1 - Z^{-1}}{1 - A·Z^{-1}} = \frac{Z^{-1}(Z - 1)}{Z^{-1}(Z - A)} \quad \implies \quad \boxed{H(Z) = \frac{Z - 1}{Z - A}} $$
